# date time notes
The +00:00 suffix is a UTC indicator

After double checking w claude: You're using the new OGC API (api.waterdata.usgs.gov) via the dataRetrieval-style waterdata.get_continuous() call — so your timestamps are UTC.

The giveaway is the USGS- prefix format in your site IDs (e.g. USGS-414527087323801) — that's the new API's monitoring location ID format. The old NWIS system used bare numbers like 414527087323801.

Where this is documented:
api.waterdata.usgs.gov/ogcapi/v0/?f=html

The time field from get_continuous() comes back as ISO 8601 UTC.
